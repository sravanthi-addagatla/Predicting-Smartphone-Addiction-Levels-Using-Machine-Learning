#### Step 3: Model Evaluation
- Regression Metrics:
    - MAE
    - RMSE
    - R2

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
def evaluate_model(model_name, y_train, train_pred, y_test, test_pred):
    
    train_mae = mean_absolute_error(y_train, train_pred)
    train_rmse = np.sqrt(mean_squared_error(y_train, train_pred))
    train_r2 = r2_score(y_train, train_pred)
    
    test_mae = mean_absolute_error(y_test, test_pred)
    test_rmse = np.sqrt(mean_squared_error(y_test, test_pred))
    test_r2 = r2_score(y_test, test_pred)
    
    print(f"\n{model_name}")
    print("-" * 40)
    print("Training:")
    print("MAE :", train_mae)
    print("RMSE:", train_rmse)
    print("R²  :", train_r2)
    
    print("\nTesting:")
    print("MAE :", test_mae)
    print("RMSE:", test_rmse)
    print("R²  :", test_r2)
```

```python
evaluate_model(
    "Linear Regression",
    y_train, linear_train_pred,
    y_test, linear_test_pred
)

evaluate_model(
    "Decision Tree",
    y_train, dt_train_pred,
    y_test, dt_test_pred
)

evaluate_model(
    "Random Forest",
    y_train, rf_train_pred,
    y_test, rf_test_pred
)

evaluate_model(
    "Gradient Boosting",
    y_train, gb_train_pred,
    y_test, gb_test_pred
)
```

#### Step 4: Overfitting & Underfitting Check
- Compare: Training vs Testing performance

```python
def get_performance(y_train, train_pred, y_test, test_pred):

    train_mae = mean_absolute_error(y_train, train_pred)
    train_rmse = np.sqrt(mean_squared_error(y_train, train_pred))
    train_r2 = r2_score(y_train, train_pred)

    test_mae = mean_absolute_error(y_test, test_pred)
    test_rmse = np.sqrt(mean_squared_error(y_test, test_pred))
    test_r2 = r2_score(y_test, test_pred)

    return [
        train_mae, train_rmse, train_r2,
        test_mae, test_rmse, test_r2
    ]
```

```python
results = []

models_predictions = {
    "Linear Regression": (
        linear_train_pred,
        linear_test_pred
    ),
    "Decision Tree": (
        dt_train_pred,
        dt_test_pred
    ),
    "Random Forest": (
        rf_train_pred,
        rf_test_pred
    ),
    "Gradient Boosting": (
        gb_train_pred,
        gb_test_pred
    )
}

for model_name, (train_pred, test_pred) in models_predictions.items():

    performance = get_performance(
        y_train,
        train_pred,
        y_test,
        test_pred
    )

    results.append([model_name] + performance)
```

```python
comparison_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Train_MAE",
        "Train_RMSE",
        "Train_R2",
        "Test_MAE",
        "Test_RMSE",
        "Test_R2"
    ]
)

comparison_df
```

```python
comparison_df["R2_Gap"] = (
    comparison_df["Train_R2"] -
    comparison_df["Test_R2"]
)

comparison_df
```

###   SPRINT 4: DEPLOYMENT 
- Goal: Make model usable in real-world applications

#### Step 1: Build ML Pipeline
- Combine: Preprocessing + Model
- Use pipeline structure


```python
print("Number of selected features:", len(selected_features))
```

```python
def create_features(df):

    df = df.copy()

    # 1. Social Media Usage Ratio
    df["Social_Media_Usage_Ratio"] = (
        df["time_on_social_media"] /
        (df["daily_usage_hours"] + 0.01)
    )

    # 2. Gaming Usage Ratio
    df["Gaming_Usage_Ratio"] = (
        df["time_on_gaming"] /
        (df["daily_usage_hours"] + 0.01)
    )

    # 3. Education Usage Ratio
    df["Education_Usage_Ratio"] = (
        df["time_on_education"] /
        (df["daily_usage_hours"] + 0.01)
    )

    # 4. Total Entertainment Time
    df["Total_Entertainment_Time"] = (
        df["time_on_social_media"] +
        df["time_on_gaming"]
    )

    # 5. Checks Per Usage Hour
    df["Checks_Per_Usage_Hour"] = (
        df["phone_checks_per_day"] /
        (df["daily_usage_hours"] + 0.01)
    )

    # 6. Weekend Usage Difference
    df["Weekend_Usage_Difference"] = (
        df["weekend_usage_hours"] -
        df["daily_usage_hours"]
    )

    # 7. Sleep Deficit
    df["Sleep_Deficit"] = (
        8 - df["sleep_hours"]
    )

    # 8. Usage vs Sleep Interaction
    df["Usage_Sleep_Interaction"] = (
        df["daily_usage_hours"] *
        df["Sleep_Deficit"]
    )

    return df
```

```python
def prepare_input(df, selected_features):

    # Create engineered features
    df = create_features(df)

    # Select exactly the features used by the model
    X = df[selected_features]

    return X
```
